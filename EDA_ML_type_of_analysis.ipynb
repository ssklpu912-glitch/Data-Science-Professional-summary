### Multivariate Analysis

```python
P_records=pd.read_csv('patient_records.csv')
P_records.head(2)
```

```python
### Scatterplot-Numerical & Numerical

plt.scatter(data=P_records, x='age', y='systolic_bp')
```

```python
sns.scatterplot(data=P_records, x='age', y='systolic_bp')
```

```python
sns.scatterplot(data=P_records, x='systolic_bp', y='heart_rate_bpm', hue='age')
```

```python
sns.scatterplot(data=P_records, x='systolic_bp', y='heart_rate_bpm', hue='age', style='smoking_status')
```

```python
sns.scatterplot(data=P_records, x='systolic_bp', y='heart_rate_bpm', hue='age', style='smoking_status', size='age')
```

```python
### Barplot-Numerical & Categorical

sns.barplot(data=P_records, x='primary_condition', y='age', palette='tab20')
plt.xticks(rotation='vertical')
```

```python
sns.barplot(data=P_records, x='insurance_type', y='age', palette='tab20')
plt.xticks(rotation='vertical')
```

```python
sns.barplot(data=P_records, x='primary_condition', y='heart_rate_bpm', palette='tab20')
plt.xticks(rotation='vertical')
```

```python
sns.barplot(data=P_records, x='primary_condition', y='heart_rate_bpm', hue='gender')
plt.xticks(rotation='vertical')
```

```python
### Box Plot --Numerical vs Categorical

sns.boxplot(data=P_records, x='primary_condition', y='heart_rate_bpm', hue='gender')
plt.xticks(rotation='vertical')
```

```python
sns.kdeplot(P_records[P_records['smoking_status']=='Never']['heart_rate_bpm'])
sns.kdeplot(P_records[P_records['smoking_status']=='Former']['heart_rate_bpm'])
```

```python
### Heatmap --Categorical vs Categorical

pd.crosstab(P_records['blood_type'], P_records['primary_condition'])
```

```python
sns.heatmap(pd.crosstab(P_records['blood_type'], P_records['primary_condition']))
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()

P_records['condition_encode'] = le.fit_transform(P_records['primary_condition'])
P_records['condition_encode'] 
```

```python
P_records.groupby('patient_id')['condition_encode'].mean()
```

```python
P_records.groupby('age')['condition_encode'].mean().plot(kind='hist')
plt.xticks(rotation='vertical')
```

```python
P_records.groupby('gender')['condition_encode'].mean().plot(kind='bar', cmap='viridis')
plt.xticks(rotation='vertical')
```

```python
### clustermap --Categorical vs Categorical

sns.clustermap(pd.crosstab(P_records['blood_type'], P_records['primary_condition']))
```
