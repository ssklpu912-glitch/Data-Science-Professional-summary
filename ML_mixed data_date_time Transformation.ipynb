### Date and time transformers

```python
Data = pd.read_csv('appointments.csv')
Data.head(2)
```

```python
Data.isnull().sum()
```

```python
Data.info()
```

### Date

```python
Data['date'] = pd.to_datetime(Data['date'])
Data['date']
```

```python
Data.info()
```

## Extract year

```python
Data['date_year'] = Data['date'].dt.year
Data.head(2)
```

## Extract month

```python
Data['date_mon'] = Data['date'].dt.month
Data.head(2)
```

```python
Data['date_mon_name'] = Data['date'].dt.month_name()
Data.head(2)
```

## extract day

```python
Data['date_day'] = Data['date'].dt.day
Data.head(2)
```

```python
Data['date_day_wk'] = Data['date'].dt.dayofweek
Data.head(2)
```

```python
Data['date_day_name'] = Data['date'].dt.day_name()
Data.head(2)
```

```python
Data['date_weekend'] = np.where(Data['date_day_name'].isin(['Sunday', 'Saturday']), 1, 0)
Data
```

## extract week of year

```python
Data['date_week'] = Data['date'].dt.isocalendar().week
Data.head(2)
```

## extract quarter

```python
Data['date_Qt'] = Data['date'].dt.quarter
Data.head(2)
```

```python
Data['date_sem'] = np.where(Data['date_Qt'].isin([1,2]), 1, 2)
Data.head(2)
```

### extract time elapsed bw dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - Data['date']
```

```python
(today - Data['date']).dt.days
```

```python
np.round((today - Data['date']) / np.timedelta64(1, 'm'), 0)
```

##  Time

```python
Data['time'] = pd.to_datetime(Data['time'], format="%H:%M")
Data['time']
```

```python
Data['time_h'] = Data['time'].dt.hour
Data.head(3)
```

```python
Data['time_m'] = Data['time'].dt.minute
Data.head(2)
```

```python
Data['time_s'] = Data['time'].dt.second
Data.head(2)
```

```python
today - Data['time']
```

```python
(today - Data['time']) / np.timedelta64(1, 's')
```

```python
(today - Data['time']) / np.timedelta64(1, 'm')
```

```python
(today - Data['time']) / np.timedelta64(1, 'h')
```
