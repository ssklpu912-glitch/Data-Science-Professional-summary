```python
###  using loop 

df=pd.DataFrame()
df
```

```python
for i in range(1, 400):           ###  for example
    res=requests.get('https://models.rcsb.org')
    temp_df=pd.DataFrame(res.json()['results']).head().[['structure', 'P_id', 'geometry']]
    df=df.append(temp_df, ignore_index=True)
```

```python
df.shape
```

```python
df.to_csv('rscb.csv')        ###  we can upload 
```
