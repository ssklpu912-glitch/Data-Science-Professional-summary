```python
## -----Problem 5-------###

### Write a program that will reverse a four digit number.Also it checks whether the reverse is true

abcd=int(input(range(1110,9999)))
reversed_num=0

while abcd != 0:
    digit = abcd % 10                                       # Get last digit
    reversed_num = reversed_num * 10 + digit               # Append digit
    abcd //= 10                                             # Remove last digit

print("The reversed number is",reversed_num)    
    
```

```python
## -----Problem 18-------###

### Write a program that will check whether the number 
##is armstrong number or not.

num_xyz=int(input("enter a three digit number"))
temp=num_xyz
ones=temp%10
tens= temp//10 %10
hundreds= temp//100

armstrong_sum=ones**3+tens**3+hundreds**3

if armstrong_sum==num_xyz:
    print("a avagadro number")
else:
    print("not a avagadro number")
```
