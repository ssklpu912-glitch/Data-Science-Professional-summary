**This study undertakes a comprehensive, multi-dataset exploratory analysis of a simulated healthcare ecosystem, examining Patient Records, Clinical Trials, Disease Outbreak, Lab Results, Medical Imaging Metadata, and Insurance, Claims both individually and in combination. At the individual level, the analysis focuses on data quality assessment, (identifying and handling missing values), descriptive and comparative statistics (BMI categorization, age-by-condition trends, treatment arm response rates, facility-level lab result variation, imaging file sizes by modality, radiologist workload, and,
    payer-level billing patterns), and time-series or categorical pattern detection (monthly disease case trends, seasonal,
    outbreak spikes, case-fatality rates by region, adverse event distribution across trial arms, and denial rates by claim**

##**converting the missing values in numerical columns in each dataset to 'median' & categorical columns to 'unknown'**

```python
Data1[Data1_num_cols].isna().sum()
```

```python
Data1_cat_cols=Data1.select_dtypes(include='str').columns
Data1[Data1_cat_cols]=Data1[Data1_cat_cols].fillna('unknown')
Data1[Data1_cat_cols]
```
