```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.impute import KNNImputer
from sklearn.compose import ColumnTransformer

from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

```python
Data = pd.read_csv('numeric_missing_target.csv')
Data.head()
```

```python
Data.isnull().sum()
```

```python
X = Data.drop(columns=['passed'])
X.head(2)
```

```python
Y = Data['passed']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.3)
X_train.head(2)
```

```python
# can do hit and trial with attribute values like neighbors, weights

kn = KNNImputer(n_neighbors = 5, weights = 'distance')

X_train_tr = kn.fit_transform(X_train)
X_test_tr = kn.fit_transform(X_test)
```

```python
pd.DataFrame(X_train_tr, columns = X_train.columns)
```

```python
clf = LogisticRegression()

clf.fit(X_train_tr, Y_train)


Y_pred = clf.predict(X_test_tr)

accuracy_score(Y_pred, Y_test)
```

## comparison with simpleImputer for accuracy check

```python
si = SimpleImputer()

X_train_trf = si.fit_transform(X_train)
X_test_trf = si.transform(X_test)
```

```python
clf2 = LogisticRegression()

clf2.fit(X_train_trf, Y_train)


Y_pred2 = clf2.predict(X_test_trf)

accuracy_score(Y_pred2, Y_test)
```
