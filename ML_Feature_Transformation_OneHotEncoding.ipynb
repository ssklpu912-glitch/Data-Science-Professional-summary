## Column Transformer

```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
P_records.head(2)
```

```python
P_records.isnull().sum()
```

```python
X=P_records[['primary_condition', 'smoking_status', 'blood_type', 'height_cm', 'systolic_bp']]
X
```

```python
Y=P_records[['heart_rate_bpm']]
Y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test=train_test_split(X, Y, test_size=0.3)
```

```python
X_train
```

```python
from sklearn.compose import ColumnTransformer
```

```python
CT=ColumnTransformer(transformers=[('tnf', OneHotEncoder(sparse_output=False, drop='first'), ['primary_condition', 'smoking_status', 'blood_type'])], remainder='passthrough')
```

```python
CT.fit_transform(X_train)
```

```python
CT.transform(X_test)
```

###  applying Column Transformer for both ordinal and nominal together

```python
P_I=pd.read_csv('lab_results.csv')
P_I.head(2)
```

```python
P_I.isnull().sum()
```

```python
X=P_I[['test_name', 'reference_range_low', 'reference_range_high', 'flag', 'ordering_facility']]
X
```

```python
Y=P_I[['result_value']]
Y
```

```python
P_I['flag'].unique()
```

```python
CT_P_I=ColumnTransformer(transformers=[
    ('tnf1', OneHotEncoder(sparse_output=False, drop='first'), ['test_name', 'ordering_facility']), 
    ('tnf2', OrdinalEncoder(categories=[['Normal', 'Low', 'High']]), ['flag']),
    ('tnf3', SimpleImputer(), ['reference_range_high'])], remainder='passthrough')
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test=train_test_split(X, Y, test_size=0.3)
```

```python
CT_P_I.fit_transform(X_train)
```

```python
CT_P_I.transform(X_test)
```
