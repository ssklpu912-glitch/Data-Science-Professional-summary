##**** applying One-Hot Encoding in Pandas--- has a limitation, it does not remember the order of columns

### applying One-Hot Encoding using sklearn class in ML

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2)
```

```python
X_train
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
OHE=OneHotEncoder(drop='first')

## drop='first'----- removes the first column from both test and train, sparse=False----  we do not need to convert to array, as by default its true
```

```python
X_train_encoded=OHE.fit_transform(X_train[['primary_condition', 'blood_type']]).toarray()
X_train_encoded
```

```python
X_test_encoded=OHE.fit_transform(X_test[['primary_condition', 'blood_type']]).toarray()
X_test_encoded
```

```python
OHE=OneHotEncoder(drop='first', sparse_output=False, dtype=np.int32)  

### dtype=np.int32 converts the values from float to int
```

```python
X_train_encoded=OHE.fit_transform(X_train[['primary_condition', 'blood_type']])
X_train_encoded
```

```python
X_test_encoded=OHE.fit_transform(X_test[['primary_condition', 'blood_type']])
X_test_encoded
```

```python
np.hstack((X_train[['height_cm', 'weight_kg']].values, X_train_encoded)).shape
```

```python
np.hstack((X_test[['height_cm', 'weight_kg']].values, X_test_encoded)).shape
```
